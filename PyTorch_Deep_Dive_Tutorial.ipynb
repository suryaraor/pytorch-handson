# 🎯 Chapter 2: Automatic Differentiation (Autograd)

This is where PyTorch really shines! Autograd automatically computes gradients for us, which is essential for training neural networks.

## 2.1 Understanding Gradients

```python
# Simple gradient example
print("🎯 Automatic Differentiation in Action")
print("=" * 50)

# Create a tensor that requires gradients
x = torch.tensor([2.0], requires_grad=True)
print(f"Input: x = {x}")
print(f"Requires grad: {x.requires_grad}")

# Define a simple function: y = x^2 + 3x + 1
y = x**2 + 3*x + 1
print(f"\nFunction: y = x² + 3x + 1")
print(f"Result: y = {y}")

# Compute gradients
y.backward()
print(f"\nGradient dy/dx = {x.grad}")
print(f"Manual calculation: dy/dx = 2x + 3 = 2(2) + 3 = 7 ✅")

print("\n" + "="*50)

# More complex example with multiple variables
print("🔗 Multiple Variable Gradients")
print("=" * 30)

# Create inputs
a = torch.tensor([3.0], requires_grad=True)
b = torch.tensor([4.0], requires_grad=True)

# Function: z = a²b + ab² (product rule example)
z = a**2 * b + a * b**2
print(f"a = {a.item()}, b = {b.item()}")
print(f"Function: z = a²b + ab²")
print(f"Result: z = {z.item()}")

# Compute gradients
z.backward()
print(f"\n∂z/∂a = {a.grad.item():.1f}")
print(f"∂z/∂b = {b.grad.item():.1f}")

# Manual verification
print(f"\nManual calculation:")
print(f"∂z/∂a = 2ab + b² = 2(3)(4) + 4² = 24 + 16 = 40 ✅")
print(f"∂z/∂b = a² + 2ab = 3² + 2(3)(4) = 9 + 24 = 33 ✅")
```

# 🧠 Chapter 3: Building Your First Neural Network

Now let's put everything together and build a simple neural network to classify data!

## 3.1 Generate Synthetic Data

```python
# Generate synthetic data for binary classification
print("📊 Generating Synthetic Data")
print("=" * 40)

# Set random seed for reproducibility
torch.manual_seed(42)

# Generate 2D data points
n_samples = 1000
X = torch.randn(n_samples, 2)  # 2D input features

# Create a circular decision boundary
# Points inside circle -> class 0, outside -> class 1
radius = 1.0
distances = torch.sqrt(X[:, 0]**2 + X[:, 1]**2)
y = (distances > radius).float()  # Convert boolean to float

print(f"Generated {n_samples} samples with 2 features")
print(f"Class distribution: {torch.bincount(y.long())}")

# Visualize the data
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
colors = ['red' if label == 0 else 'blue' for label in y]
plt.scatter(X[:, 0], X[:, 1], c=colors, alpha=0.6, s=20)
circle = plt.Circle((0, 0), radius, fill=False, color='black', linestyle='--', linewidth=2)
plt.gca().add_patch(circle)
plt.title('Synthetic Dataset\n(Red: Class 0, Blue: Class 1)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.axis('equal')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(y.numpy(), bins=2, alpha=0.7, color=['red', 'blue'])
plt.title('Class Distribution')
plt.xlabel('Class')
plt.ylabel('Count')
plt.xticks([0, 1])

plt.tight_layout()
plt.show()

print(f"Data shape: {X.shape}")
print(f"Labels shape: {y.shape}")
print(f"Features range: [{X.min():.2f}, {X.max():.2f}]")
```

## 3.2 Define the Neural Network

Let's create a simple neural network with one hidden layer.

```python
# Define a simple neural network class
class SimpleNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        
        # Define layers
        self.hidden = nn.Linear(input_size, hidden_size)  # Hidden layer
        self.output = nn.Linear(hidden_size, output_size)  # Output layer
        self.activation = nn.ReLU()  # Activation function
        
    def forward(self, x):
        # Forward pass through the network
        x = self.hidden(x)      # Linear transformation
        x = self.activation(x)  # Apply ReLU activation
        x = self.output(x)      # Final linear transformation
        return x

# Create the network
input_size = 2    # 2D input features
hidden_size = 10  # 10 neurons in hidden layer
output_size = 1   # 1 output for binary classification

model = SimpleNN(input_size, hidden_size, output_size)
print("🧠 Neural Network Architecture")
print("=" * 40)
print(model)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"\nTotal parameters: {total_params}")
print(f"Trainable parameters: {trainable_params}")

# Move model to device
model = model.to(device)
print(f"Model moved to: {device}")

# Test the model with a sample input
with torch.no_grad():
    sample_input = torch.randn(5, 2).to(device)  # 5 samples, 2 features
    sample_output = model(sample_input)
    print(f"\nTest input shape: {sample_input.shape}")
    print(f"Test output shape: {sample_output.shape}")
    print(f"Sample outputs: {sample_output.squeeze()}")
```

## 3.3 Training the Network

Now let's train our neural network! This involves several steps:
1. Define a loss function
2. Choose an optimizer 
3. Run the training loop

```python
# Prepare data for training
X_train = X.to(device)
y_train = y.to(device)

# Define loss function and optimizer
criterion = nn.BCEWithLogitsLoss()  # Binary Cross Entropy with Logits
optimizer = optim.Adam(model.parameters(), lr=0.01)

print("🚀 Training Neural Network")
print("=" * 40)

# Training loop
epochs = 100
losses = []

start_time = time.time()

for epoch in range(epochs):
    # Forward pass
    outputs = model(X_train).squeeze()
    loss = criterion(outputs, y_train)
    
    # Backward pass
    optimizer.zero_grad()  # Clear gradients
    loss.backward()        # Compute gradients
    optimizer.step()       # Update weights
    
    # Store loss for plotting
    losses.append(loss.item())
    
    # Print progress
    if (epoch + 1) % 20 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')

training_time = time.time() - start_time
print(f"\n✅ Training completed in {training_time:.2f} seconds")

# Plot training loss
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(losses)
plt.title('Training Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, alpha=0.3)

# Evaluate the model
model.eval()
with torch.no_grad():
    predictions = torch.sigmoid(model(X_train).squeeze())
    predicted_classes = (predictions > 0.5).float()
    accuracy = (predicted_classes == y_train).float().mean()

print(f"Final Training Accuracy: {accuracy:.2%}")

# Create decision boundary visualization
plt.subplot(1, 2, 2)

# Create a mesh of points
h = 0.02
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_min, y_max, h))

# Make predictions on the mesh
mesh_points = torch.tensor(np.c_[xx.ravel(), yy.ravel()], dtype=torch.float32).to(device)
with torch.no_grad():
    Z = torch.sigmoid(model(mesh_points).squeeze())
    Z = Z.reshape(xx.shape).cpu().numpy()

# Plot the decision boundary
plt.contourf(xx, yy, Z, levels=50, alpha=0.8, cmap=plt.cm.RdYlBu)
plt.colorbar()

# Plot the data points
colors = ['red' if label == 0 else 'blue' for label in y]
plt.scatter(X[:, 0], X[:, 1], c=colors, alpha=0.6, s=20)
plt.title(f'Decision Boundary (Accuracy: {accuracy:.1%})')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

plt.tight_layout()
plt.show()
```
